```python
# let us create a new datatype using python magic fuction for fraction

class fraction():

    #cosntructor magic function 1
    def __init__(self,x,y):
        self.num = x
        self.den = y
        
    def __str__(self):   #this fuction used to display our object solution as we want
        return '{}/{}'.format(self.num,self.den)

    def __add__(self,other):   #when two fraction are added for two different objects
        new_num = self.num*other.den + other.num*self.den
        new_den = self.den*other.den
        return '{}/{}'.format(new_num,new_den)

    def __sub__(self,other):   #when two fraction are substracted for two different objects
        new_num = self.num*other.den - other.num*self.den
        new_den = self.den*other.den
        return '{}/{}'.format(new_num,new_den)

    def __mul__(self,other):   #when two fraction are multiplied for two different objects
        new_num = self.num*other.num 
        new_den = self.den*other.den
        return '{}/{}'.format(new_num,new_den)

    def __truediv__(self,other):   #when two fraction are divided for two different objects
        new_num = self.num*other.den 
        new_den = self.den*other.num
        return '{}/{}'.format(new_num,new_den)

    def floatDiv(self):
        return self.num/self.den
```
