```python
class ATM:
    def __init__(self):  # constructor
        self.pin = ""
        self.balance = 0
        self.menu()

    def menu(self):
        user_inp = input(""" Hello sir/mam, what would you like to do? 
        
                        1. Press 1 to create user pin
                        2. Press 2 to change user pin
                        3. Press 3 to Check Balance
                        4. Press 4 to Withdraw amount
                        5. Press Anything else to Exit: """)

        if user_inp == '1':
            # create pin
            self.create_pin()

        elif user_inp == '2':
            # change pin
            self.change_pin()

        elif user_inp == '3':
            # check balance
            self.check_balance()
            

        elif user_inp == '4':
            # withdraw
            self.withdraw()

        else:
            exit()
    def create_pin(self):
        user_pin = int(input("Entre your pin"))
        self.pin = user_pin

        user_balance = int(input("Entre user balacne"))
        self.balance = user_balance
        print("Pin created succesfully")
        self.menu()

    def change_pin(self):
        user_pin = int(input("Entre your pin"))
        if user_pin == self.pin:
            user_pin = int(input("Entre new pin"))
            self.pin = user_pin
            print("PIN changed succefully")
        else:
            print("Incorrect pin entred")
            
        self.menu()

    def check_balance(self):
        user_pin = int(input("entre your pin"))
        if user_pin == self.pin:
            print("Your Balance is",self.balance)
        else :
            print("Incorrect pin entred")
        self.menu()

    def withdraw(self):
        amount = int(input("Entre your amount to withdraw"))
        if amount <= self.balance :
            self.balance = self.balance - amount
            print("Amount withdrawed")
        else:
            print("Insufficient Balance")

        self.menu()

        
```
